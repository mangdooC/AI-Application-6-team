```python
import pandas as pd
import glob

file_paths = glob.glob('/content/drive/MyDrive/project2025/ErrorIndex/*.xls')
data_frames = [] # 데이터를 담을 리스트 생성

for path in file_paths:
    df = pd.read_excel(path, usecols='B:M')

    # 전체 문장에서 교정할 거리가 있는 경우만 필터링
    filtered_df = df[df['교정 형태소'].notna() & ~df.apply(lambda row: all(str(cell) in ['문장', '원 형태소', '교정 형태소'] for cell in row), axis=1)][['문장', '원 형태소', '교정 형태소']]

    # 필터링된 행들을 리스트에 추가
    data_frames.append(filtered_df)

# 모든 필터된 DataFrame을 하나로 합치기
result_df = pd.concat(data_frames, ignore_index=True)
# 결과 출력
for row in data_frames:
    print(row)
```
